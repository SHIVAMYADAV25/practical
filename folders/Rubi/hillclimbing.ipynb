```python
points = [(1,5) , (6,4),(5,2),(2,1)]

x , y = 1,1
step = 0.1

def total_distance(x,y):
    return sum((x - px) ** 2 + (y - py) ** 2 for px , py in points)

current = total_distance(x,y);
iteration = 1

while True:
    print(f"Iteration : {iteration} : ({x:.2f} , {y:.2f}) Distance : {current:.2f}")

    neighbour = [(x+step,y) , (x-step,y) , (x,y+step) , (x,y-step)]

    best_x , best_y = x , y
    best_dist = current

    for nx , ny in neighbour:
        d = total_distance(nx,ny)
        if d < best_dist:
            best_x , best_y = nx, ny
            best_dist = d

    if best_dist >= current:
        break

    current = best_dist
    x , y = best_x , best_y
    iteration+=1


print("\n Optimimal point : ", (round(x,2) , round(y,2)))
print("\n Minimum Distance : ",round(current,2))
      
```
