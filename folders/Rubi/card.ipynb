```python
import random
```

```python
suits = ['heart','spades','clube','diamond']
ranks = ['2','3','4','5','6','7','8','9','J','Q','K','A']
```

```python
deck = []
```

```python
for suit in suits:
    for rank in ranks:
        deck.append(rank + " ok " + suit);
```

```python
print(deck)
```

```python
random.shuffle(deck)
```
