```python
from collections import deque


def is_valid(missionaries_left, cannibals_left):

    # Check left side
    if missionaries_left < 0 or cannibals_left < 0:
        return False

    if missionaries_left > 3 or cannibals_left > 3:
        return False

    if missionaries_left > 0:
        if missionaries_left < cannibals_left:
            return False

    # Check right side
    missionaries_right = 3 - missionaries_left
    cannibals_right = 3 - cannibals_left

    if missionaries_right > 0:
        if missionaries_right < cannibals_right:
            return False

    return True


def solve():

    # (missionaries on left, cannibals on left, boat position)
    # 0 = boat on left
    # 1 = boat on right

    start_state = (3, 3, 0)
    goal_state = (0, 0, 1)

    # Possible people that can travel in boat
    possible_moves = [
        (1, 0),   # 1 missionary
        (2, 0),   # 2 missionaries
        (0, 1),   # 1 cannibal
        (0, 2),   # 2 cannibals
        (1, 1)    # 1 missionary and 1 cannibal
    ]

    queue = deque()

    queue.append((start_state, [start_state]))

    visited = {start_state}

    while queue:

        current_state, path = queue.popleft()

        missionaries_left = current_state[0]
        cannibals_left = current_state[1]
        boat_position = current_state[2]

        # Goal reached
        if current_state == goal_state:
            return path

        # Try every possible boat move
        for missionaries_moving, cannibals_moving in possible_moves:

            # Boat is on left side
            if boat_position == 0:

                new_missionaries_left = (
                    missionaries_left - missionaries_moving
                )

                new_cannibals_left = (
                    cannibals_left - cannibals_moving
                )

                new_boat_position = 1

            # Boat is on right side
            else:

                new_missionaries_left = (
                    missionaries_left + missionaries_moving
                )

                new_cannibals_left = (
                    cannibals_left + cannibals_moving
                )

                new_boat_position = 0

            new_state = (
                new_missionaries_left,
                new_cannibals_left,
                new_boat_position
            )

            # Check if the new state is valid
            if is_valid(
                new_missionaries_left,
                new_cannibals_left
            ):

                # Check if we already visited this state
                if new_state not in visited:

                    visited.add(new_state)

                    new_path = path + [new_state]

                    queue.append(
                        (new_state, new_path)
                    )

    return None


# Run the program
solution = solve()


# Print the solution
print("MISSIONARIES AND CANNIBALS")
print("--------------------------")

for step, state in enumerate(solution):

    missionaries_left = state[0]
    cannibals_left = state[1]
    boat_position = state[2]

    if boat_position == 0:
        boat = "Left"
    else:
        boat = "Right"

    print(
        "Step", step,
        ":", 
        "Missionaries =", missionaries_left,
        "Cannibals =", cannibals_left,
        "Boat =", boat
    )
```

```python
from collections import deque


# Check if the state is safe
def valid(m, c):

    # Left side is safe
    left_side = (m == 0 or m >= c)

    # Right side is safe
    right_m = 3 - m
    right_c = 3 - c

    right_side = (right_m == 0 or right_m >= right_c)

    return 0 <= m <= 3 and 0 <= c <= 3 and left_side and right_side


def solve():

    # (missionaries, cannibals, boat)
    # 0 = boat on left
    # 1 = boat on right

    start = (3, 3, 0)
    goal = (0, 0, 1)

    # Possible boat moves
    moves = [
        (1, 0),  # 1 missionary
        (2, 0),  # 2 missionaries
        (0, 1),  # 1 cannibal
        (0, 2),  # 2 cannibals
        (1, 1)   # 1 missionary + 1 cannibal
    ]

    # Queue
    queue = deque()

    # Put starting state in queue
    queue.append((start, [start]))

    # Remember visited states
    visited = {start}


    while queue:

        # Take first state from queue
        state, path = queue.popleft()

        m, c, b = state


        # If goal reached
        if state == goal:
            return path


        # Try every boat move
        for dm, dc in moves:

            # Boat is on left
            if b == 0:

                new_state = (
                    m - dm,
                    c - dc,
                    1
                )

            # Boat is on right
            else:

                new_state = (
                    m + dm,
                    c + dc,
                    0
                )


            # Check new state
            if valid(new_state[0], new_state[1]):

                # Check if already visited
                if new_state not in visited:

                    visited.add(new_state)

                    # Add new state to queue
                    queue.append(
                        (new_state, path + [new_state])
                    )


    return None


# Run the program
path = solve()


# Print solution
print("Solution:")

for i, state in enumerate(path):

    print("Step", i, state)
```
