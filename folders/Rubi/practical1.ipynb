```python
from collections import deque
```

```python
# this is jug capacaity
jug1_capacity = 4
jug2_capacity = 3
```

```python
initial_state = (0,0)
```

```python
# joo bhi state ko hume process kar leya hai usai hum ismai dal deete hai
visited = set()
```

```python
# created queue initailly ismai sirf (0,0)
queue = deque([initial_state])
```

```python
print("state Space:")
```

```python
while queue:
    state = queue.popleft() # takes the first value which will be (0,0)

    if state in visited:
        continue

    visited.add(state)

    x,y = state

    print(state)

    # All possible states I can reach from the current state.
    next_states=[
        (jug1_capacity,y), # fill jug 1
        (x,jug2_capacity), # fill jug 2
        (0,y), # empty jug 1 (make the jug 1 zero and keep the jug2 as it is)
        (x,0)  # empty jug 2 (opposite)
    ]

    # 1. Fill Jug 1
    # 2. Fill Jug 2
    # 3. Empty Jug 1
    # 4. Empty Jug 2
    # 5. Pour Jug 1 → Jug 2
    # 6. Pour Jug 2 → Jug 1

    # check how much capacity the jug1 can pour into jug2)
    #transfer = minimum( water available in Jug 1, empty space in Jug 2 )
    max_transfer = min(x,jug2_capacity - y)
    next_states.append((x-max_transfer,y+max_transfer)) #This represents transferring water from Jug 1 to Jug 2

    # transfer jug2 to jug1
    transfer = min(y,jug1_capacity-x)
    next_states.append((x+transfer,y-transfer))

    for next_state in next_states:
        if next_state not in visited:
            queue.append(next_state);
            
```

```python
print("\nToatal reachable states =",len(visited))
```
