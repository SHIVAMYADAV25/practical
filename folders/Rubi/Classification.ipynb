```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    log_loss,
    accuracy_score,
    confusion_matrix,
    recall_score,
    precision_score,
    f1_score
)
```

```python
x = np.array([[1],[2],[3],[4],[5],[6],[7],[8]])

y = np.array([0,0,0,0,1,1,1,1])
```

```python
model = LogisticRegression();
```

```python
model.fit(x,y);
```

```python
y_pred = model.predict(x)
```

```python
y_prob = model.predict_proba(x)
```

```python
cost = log_loss(y,y_prob)
print("Log Loss(Cost Function):",cost)

# how good the prediction probability are
```

```python
accuracy = accuracy_score(y,y_pred)
print("Accuracy:",accuracy)
#How many predictions were correct out of all predictions.
```

```python
cm = confusion_matrix(y,y_pred);
print("\nConfusion Matrix:")
print(cm)
```

```python
#Precision
precision = precision_score(y,y_pred)
print("Precision:",precision)
#Of all the things the model predicted as Positive, how many were actually Positive?
```

```python
recall = recall_score(y,y_pred)
print(recall)
#Out of all actual Positive cases, how many did I find?
```

```python
f1 = f1_score(y,y_pred)
print(f1)
```
