```python
class QueenChessBoard:

    def __init__(self, size):
        self.size = size
        self.columns = []

    def place_in_next(self, column):
        self.columns.append(column)

    def remove_in_current_row(self):
        return self.columns.pop()

    def is_column_safe(self, column):
        row = len(self.columns)

        # Check same column
        for queen_column in self.columns:
            if column == queen_column:
                return False

        # Check left-to-right diagonal
        for queen_row, queen_column in enumerate(self.columns):
            if queen_column - queen_row == column - row:
                return False

        # Check right-to-left diagonal
        for queen_row, queen_column in enumerate(self.columns):
            if queen_column + queen_row == column + row:
                return False

        return True

    def display(self):
        for row in range(self.size):
            for column in range(self.size):
                if column == self.columns[row]:
                    print("Q", end=" ")
                else:
                    print("-", end=" ")
            print()

    def solve_queen(self):
        number_of_solutions = 0

        row = 0
        column = 0

        while True:

            # Try to find a safe column in the current row
            while column < self.size:

                if self.is_column_safe(column):
                    self.place_in_next(column)
                    row += 1
                    column = 0
                    break

                column += 1

            # A complete solution has been found
            if row == self.size:
                self.display()
                print()

                number_of_solutions += 1

                # Backtrack to find another solution
                previous_column = self.remove_in_current_row()
                row -= 1
                column = previous_column + 1
                continue

            # No safe column was found in the current row
            if column == self.size:

                # If we are at the first row, there are no more solutions
                if row == 0:
                    break

                # Backtrack one row
                previous_column = self.remove_in_current_row()
                row -= 1
                column = previous_column + 1

        print("Number of Solutions:", number_of_solutions)


size = int(input("Enter the number of queens: "))

board = QueenChessBoard(size)
board.solve_queen()

#O(N!)
#O(N)

# question sir bole ga ki 16 dalu as input then why it is talking time and other thing also the 16^2 hoga all that thing in the code
```
